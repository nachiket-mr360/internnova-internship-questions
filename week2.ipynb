Task 7: Handling Missing Values 

Take a dataset containing missing values and perform the following:

Identify missing values using isnull() / isna().
Count missing values in each column.
Remove rows containing missing values.
Fill missing values using appropriate methods.
Display the dataset before and after handling missing values.

Briefly explain why handling missing data is important in Data Analytics

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Rahul", "Priya", "Amit", "Sneha", "Karan", "Neha"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", None],
    "Age": [25, 28, np.nan, 30, 27, 26],
    "Salary": [50000, 45000, 60000, np.nan, 48000, 70000],
    "Experience": [2, 4, 3, 5, np.nan, 4]
}

df = pd.DataFrame(data)
df_original = df
print(f'identifing missing values:\n{df.isnull}')
print(f'classifying missing values:\n{df.isnull().sum()}')
print(f'removing rows contains missing values:\n{df.dropna(inplace=False)}')#for permanent change use inplace=True
print(f'filling missing values:')
print(f'filling Age with mean\n{df['Age'].fillna(df["Age"].mean(),inplace=True)}')
print(f'filling salary with mean\n{df['Salary'].fillna(df['Salary'].mean(),inplace=True)}')
print(f'filling Experience with median\n{df['Experience'].fillna(df['Experience'].median(),inplace=True)}')
print(f'filling department using na:\n{df['Department'].fillna('Na',inplace=True)}')
print(f'original dataset:\n{df_original}')
print(f'updated dataset:\n{df}')

```

Task 8: Merge, Concatenate, GroupBy & Pivot Table 

Using two or more datasets, perform the following operations:

Merge & Concatenate
Merge two DataFrames using a common column.
Concatenate two DataFrames.
GroupBy
Group data based on a selected column.
Calculate aggregate values such as sum, mean, count, minimum, or maximum.
Pivot Table
Create a Pivot Table using Pandas.
Summarize the data based on appropriate rows, columns, and values.

```python

import pandas as pd

employees = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Rahul", "Priya", "Amit", "Sneha", "Karan"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [50000, 45000, 60000, 55000, 48000]
})

locations = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104, 105],
    "City": ["Bangalore", "Mumbai", "Bangalore", "Pune", "Mumbai"],
    "Experience": [2, 4, 3, 5, 2]
})
merged_df = pd.merge(
    employees,
    locations,
    on="Employee_ID",
    how="inner"
)

print("Merged DataFrame:")
print(merged_df)
new_employees = pd.DataFrame({
    "Employee_ID": [106, 107],
    "Name": ["Neha", "Arjun"],
    "Department": ["IT", "Finance"],
    "Salary": [70000, 52000]
})

concatenated_df = pd.concat(
    [employees, new_employees],
    ignore_index=True
)

print("\nConcatenated DataFrame:")
print(concatenated_df)


grouped = employees.groupby("Department")["Salary"].sum()

print("\nTotal Salary by Department:")
print(grouped)


grouped_stats = employees.groupby("Department")["Salary"].agg(
    ["sum", "mean", "count", "min", "max"]
)

print("\nSalary Statistics by Department:")
print(grouped_stats)

pivot = pd.pivot_table(
    merged_df,
    values="Salary",
    index="Department",
    columns="City",
    aggfunc="mean"
)

print("\nPivot Table:")
print(pivot)
```
